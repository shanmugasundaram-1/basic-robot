```python
from jetbot import Robot
import time

robot = Robot()

SPEED_SLOW = 0.20
SPEED_FAST = 0.35

LEFT_LOOP_TIME = 3.0
RIGHT_LOOP_TIME = 3.0
PAUSE_TIME = 0.3

try:
    while True:
        robot.set_motors(SPEED_SLOW, SPEED_FAST)
        time.sleep(LEFT_LOOP_TIME)

        # Brief transition at the centre
        robot.stop()
        time.sleep(PAUSE_TIME)

        # Second loop: curve right
        robot.set_motors(SPEED_FAST, SPEED_SLOW)
        time.sleep(RIGHT_LOOP_TIME)

        # Repeat the figure-8 pattern
        robot.stop()
        time.sleep(PAUSE_TIME)

except KeyboardInterrupt:
    print("Robot stopped by user")

finally:
    robot.stop()
```
